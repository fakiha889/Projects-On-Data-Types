# **Task3 : Streaming Data**

# **Description**
 **I wanted to simulate a stream of temperature readings, display the time of each reading, and generate an alert whenever the temperature exceeded 35°C.**

```python
import time
from datetime import datetime

temperatures = [25, 28, 32, 40, 27]

for temp in temperatures:
    current_time = datetime.now().strftime("%H:%M:%S")

    print(current_time, "Temperature:", temp)

    if temp > 35:
        print("ALERT: High Temperature!")
    else:
        print("Temperature is Normal")

    time.sleep(1)
```

# **Task5 : Unstructured and Natural Language Data**
# **Description**
**In this task, I analyzed customer feedback reviews using Python. I searched for positive and negative keywords and counted how many reviews were positive and negative.**

```python
import pandas as pd
reviews = ["The customer service was excellent and very fast!",
           "App crashed during bill payment. Extremely frustrating experience.",
            "Great interface, easy to navigate, highly recommended.",
            "Worst support ever. Nobody responded to my complaint for three days."
            ]
positive_kwords = ["excellent","great","remommended","easy"]
negative_kwords = ["crashed","frustating","worse","complaint"]
pos_count = 0
neg_count = 0
for i, review in enumerate(reviews,1):
  words = review.lower().split()
  is_positive = any(kw in words or kw in review.lower() for kw in positive_kwords)
  is_negative = any(kw in words or kw in review.lower() for kw in negative_kwords)
  sentiment = "Positive" if is_positive else ("Negative" if is_negative else
"Neutral")
if is_positive: pos_count += 1
if is_negative: neg_count += 1
print(f"Review {i}: [{sentiment}] '{review}'")
print(f"\nSummary Sentiment Counts -> Positive: {pos_count}, Negative: {neg_count}")
```

# **Task6 : Machine-Generated Data**
# **Description**
**In this task, I worked with temperature sensor readings. I used Pandas to organize the readings and identify temperatures greater than 75°C as unusual readings.**

```python
import pandas as pd
log_data = {
'Timestamp': ['08:00:01', '08:00:05', '08:00:10', '08:00:15',
'08:00:20'],
'SensorID': ['TEMP_01', 'TEMP_01', 'TEMP_01', 'TEMP_01', 'TEMP_01'],
'Temp_Celsius': [62.4, 63.1, 78.9, 82.5, 64.0],
'Status': ['OK', 'OK', 'WARNING', 'CRITICAL', 'OK']
}
logs_df = pd.DataFrame(log_data)
anomalies = logs_df[logs_df['Temp_Celsius'] > 75.0]
print("Full IoT Sensor Log Stream:\n", logs_df)
print("\n[ALERT] Overheat Sensor Anomalies Detected:\n", anomalies)
```
